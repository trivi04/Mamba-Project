```python
import torch
import time
import math
import matplotlib.pyplot as plt
import pandas as pd
from transformers import AutoTokenizer, AutoModelForCausalLM, MambaForCausalLM

device = "cuda" if torch.cuda.is_available() else "cpu"

sequence_lengths = [256, 512, 1024, 2048]
batch_sizes = [1, 4]
results = []

def compute_perplexity(model, inputs):
    with torch.no_grad():
        outputs = model(**inputs, labels=inputs["input_ids"])
        loss = outputs.loss.item()
    return math.exp(loss)


def benchmark_model(model, tokenizer, name, sequence_lengths, batch_sizes):
    model.to(device)
    model.eval()

    if tokenizer.pad_token is None:
        tokenizer.pad_token = tokenizer.eos_token

    for seq_len in sequence_lengths:
        for batch_size in batch_sizes:
            text = "The quick brown fox jumps over the lazy dog. " * (seq_len // 8)
            inputs = tokenizer(
                [text] * batch_size,
                return_tensors="pt",
                padding=True,
                truncation=True,
            ).to(device)

            if "attention_mask" not in inputs:
                inputs["attention_mask"] = torch.ones_like(inputs["input_ids"])

            # Warm-up
            with torch.no_grad():
                _ = model(**inputs)

            torch.cuda.synchronize()
            torch.cuda.reset_peak_memory_stats()

            start = time.perf_counter()
            with torch.no_grad():
                _ = model(**inputs)
            torch.cuda.synchronize()
            end = time.perf_counter()

            elapsed = end - start
            mem = torch.cuda.max_memory_allocated() / 1024**2

            ppl = compute_perplexity(model, inputs)

            results.append({
                "model": name,
                "seq_len": seq_len,
                "batch": batch_size,
                "time": elapsed,
                "mem": mem,
                "perplexity": ppl
            })

            print(f"{name} | Seq={seq_len:4d} | Batch={batch_size:2d} | Time={elapsed:.4f}s | Mem={mem:.1f} MB | PPL={ppl:.2f}")


# GPT-2
gpt2_tokenizer = AutoTokenizer.from_pretrained("gpt2")
gpt2_model = AutoModelForCausalLM.from_pretrained("gpt2")
benchmark_model(gpt2_model, gpt2_tokenizer, "GPT-2", sequence_lengths, batch_sizes)





# Mamba 130M
mamba_tokenizer = AutoTokenizer.from_pretrained("state-spaces/mamba-130m-hf")
mamba_model = MambaForCausalLM.from_pretrained("state-spaces/mamba-130m-hf")
benchmark_model(mamba_model, mamba_tokenizer, "Mamba 130M", sequence_lengths, batch_sizes)





df = pd.DataFrame(results)

# Inference time
for seq in sequence_lengths:
    subset = df[df["seq_len"] == seq]
    plt.figure(figsize=(7,4))
    for model_name in subset["model"].unique():
        model_data = subset[subset["model"] == model_name]
        plt.plot(model_data["batch"], model_data["time"], marker='o', label=model_name)
    plt.xlabel("Batch size")
    plt.ylabel("Inference time (s)")
    plt.title(f"Inference Time — Seq Length {seq}")
    plt.legend()
    plt.grid(True)
    plt.show()

# Perplexity comparison
plt.figure(figsize=(7,4))
for model_name in df["model"].unique():
    model_data = df[df["model"] == model_name]
    plt.plot(model_data["seq_len"], model_data["perplexity"], marker='s', label=model_name)
plt.xlabel("Sequence Length")
plt.ylabel("Perplexity (↓ better)")
plt.title("Perplexity vs Sequence Length")
plt.legend()
plt.grid(True)
plt.show()

```

```python
!pip install -q transformers datasets torch matplotlib


import torch, time, gc
from datasets import load_dataset
from transformers import AutoTokenizer, AutoModelForCausalLM
import pandas as pd


device = "cuda" if torch.cuda.is_available() else "cpu"
print("Running Transformer on:", device)


dataset = load_dataset("wikitext", "wikitext-2-raw-v1", split="train[:1%]")
texts = dataset["text"][:4]


tokenizer = AutoTokenizer.from_pretrained("sshleifer/tiny-gpt2")
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token


def encode(texts, max_len):
    return tokenizer(texts,
                     truncation=True,
                     padding="max_length",
                     max_length=max_len,
                     return_tensors="pt")


model_name = "sshleifer/tiny-gpt2"
model = AutoModelForCausalLM.from_pretrained(model_name, low_cpu_mem_usage=True)
model.eval()


def benchmark(model, inputs, device):
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats()
    gc.collect()


    # move inputs to device
    inputs = {k: v.to(device) for k, v in inputs.items()}
    model.to(device)


    start = time.time()
    with torch.no_grad():
        if device.startswith("cuda"):
            with torch.autocast(device_type="cuda", dtype=torch.float16):
                _ = model(**inputs)
        else:
            _ = model(**inputs)
    torch.cuda.synchronize() if device.startswith("cuda") else None
    end = time.time()


    mem = torch.cuda.max_memory_allocated() / 1e6 if device.startswith("cuda") else 0
    model.to("cpu")
    torch.cuda.empty_cache()
    return end - start, mem


seq_lengths = [256, 512, 1024]
results = []
for L in seq_lengths:
    print(f"\nSeq={L}")
    inputs = encode(texts, L)
    try:
        t, m = benchmark(model, inputs, device)
        results.append([L, t, m])
        print(f"Transformer: time={t:.3f}s mem={m:.1f}MB")
    except RuntimeError as e:
        print("Error (Transformer):", e)
        torch.cuda.empty_cache()
        break


df = pd.DataFrame(results, columns=["SeqLen", "Time_s", "Mem_MB"])
df.to_csv("transformer_results_safe.csv", index=False)
print("\nSaved transformer_results_safe.csv")
print(df)
```

# Attention (O(N^2)) vs Prefix-linear (O(N)) benchmark

```python

# Colab-friendly: runs with small hidden dims and small batches; no big downloads.

!pip install -q matplotlib psutil pandas

import torch, time, gc, psutil
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", device)

# Settings
seq_lengths = [128, 256, 512, 1024, 2048]   # remove 2048 if OOMs
hidden_dim = 64
batch = 1


```

```python
def bench_attention(seq_len, dtype=torch.float32):
    # Allocate Q, K, V
    q = torch.randn(batch, seq_len, hidden_dim, dtype=dtype, device=device)
    k = torch.randn(batch, seq_len, hidden_dim, dtype=dtype, device=device)
    v = torch.randn(batch, seq_len, hidden_dim, dtype=dtype, device=device)

    if device == "cuda":
        torch.cuda.empty_cache()
        torch.cuda.reset_peak_memory_stats()
    gc.collect()

    torch.cuda.synchronize() if device=="cuda" else None
    t0 = time.time()

    # Full attention computation
    scores = torch.matmul(q, k.transpose(-2, -1)) / (hidden_dim**0.5)
    weights = torch.softmax(scores, dim=-1)
    out = torch.matmul(weights, v)

    torch.cuda.synchronize() if device=="cuda" else None
    t1 = time.time()

    mem = torch.cuda.max_memory_allocated() / 1e6 if device=="cuda" else psutil.Process().memory_info().rss / 1e6
    return (t1 - t0), mem


```

```python
def bench_prefix_linear(seq_len, dtype=torch.float32):
    # Simulate O(N) prefix computation
    x = torch.randn(batch, seq_len, hidden_dim, dtype=dtype, device=device)

    if device == "cuda":
        torch.cuda.empty_cache()
        torch.cuda.reset_peak_memory_stats()
    gc.collect()

    torch.cuda.synchronize() if device=="cuda" else None
    t0 = time.time()

    # Per-token projection + cumulative sum (O(N))
    per = x
    prefix = torch.cumsum(per, dim=1)
    out = prefix

    torch.cuda.synchronize() if device=="cuda" else None
    t1 = time.time()

    mem = torch.cuda.max_memory_allocated() / 1e6 if device=="cuda" else psutil.Process().memory_info().rss / 1e6
    return (t1 - t0), mem


# Run benchmarks
att_times, att_mems = [], []
lin_times, lin_mems = [], []
worked_lengths = []

for L in seq_lengths:
    try:
        print(f"Running seq_len={L} ...", end=" ")
        a_t, a_m = bench_attention(L, dtype=torch.float16 if device=="cuda" else torch.float32)
        l_t, l_m = bench_prefix_linear(L, dtype=torch.float16 if device=="cuda" else torch.float32)
        print(f"ATT time={a_t:.4f}s mem={a_m:.1f}MB | LIN time={l_t:.4f}s mem={l_m:.1f}MB")
        att_times.append(a_t); att_mems.append(a_m)
        lin_times.append(l_t); lin_mems.append(l_m)
        worked_lengths.append(L)
    except RuntimeError as e:
        print("OOM or error:", e)
        break


```

```python
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.plot(worked_lengths, att_times, 'o-', label='Attention (O(N^2))')
plt.plot(worked_lengths, lin_times, 'o-', label='Prefix-linear (O(N))')
plt.xlabel('Sequence length')
plt.ylabel('Time (s)')
plt.title('Time scaling')
plt.legend()
plt.grid(True)

plt.subplot(1,2,2)
plt.plot(worked_lengths, att_mems, 'o-', label='Attention mem (MB)')
plt.plot(worked_lengths, lin_mems, 'o-', label='Prefix-linear mem (MB)')
plt.xlabel('Sequence length')
plt.ylabel('Memory (MB)')
plt.title('Memory scaling')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()

df = pd.DataFrame({
    'SeqLen': worked_lengths,
    'Attention_Time_s': att_times,
    'Prefix_Time_s': lin_times,
    'Attention_Mem_MB': att_mems,
    'Prefix_Mem_MB': lin_mems
})
print("\nResults:\n", df.to_string(index=False))
```
